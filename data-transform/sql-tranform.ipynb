```python
import pandas as pd
import numpy as np
import os
import sys
```

```python
df = pd.read_csv('/Users/prasadpilankar/Documents/BAN/DE2025/Comtrade-Analytics/data-cleaning/cleandata_trade.csv')
df.head()
```

```python
df.columns
```

```python
selected_columns = [
    'typecode','refyear', 'reporteriso', 'reporterdesc',
    'partneriso', 'partnerdesc',
    'flowcode', 'cmdcode', 'cmddesc',
    'qty', 'netwgt', 'cifvalue', 'fobvalue','primaryvalue','isreported'
]
```

```python
# Select only the relevant columns
df_selected = df[selected_columns]
df_selected.head()
```

```python
df_selected['primaryvalue'] = pd.to_numeric(df_selected['primaryvalue'], errors='coerce').fillna(0).astype('int64')
df_selected['fobvalue'] = pd.to_numeric(df_selected['fobvalue'], errors='coerce').fillna(0).astype('int64')
df_selected['cifvalue'] = pd.to_numeric(df_selected['cifvalue'], errors='coerce').fillna(0).astype('int64')
df_selected.head()
```

```python
df_selected.info()
```

```python
cols_to_convert = ['reporteriso', 'reporterdesc', 'partneriso', 'partnerdesc', 'flowcode', 'cmdcode', 'cmddesc']
df_selected[cols_to_convert] = df_selected[cols_to_convert].astype('string')
df_selected.info()
```

```python
df_selected.to_csv('/Users/prasadpilankar/Documents/BAN/DE2025/Comtrade-Analytics/data-transform/cleandata_trade_cleaned.csv', index=False)
```
